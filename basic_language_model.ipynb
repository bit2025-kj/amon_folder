```python
def window(iterable, n):
    """
    Returns a sliding window (of width n) over data from the iterable.
    s -> (s0,s1,...s[n-1]), (s1,s2,...,sn), ...
    """
    from collections import deque
    from itertools import islice

    it = iter(iterable)
    result = tuple(islice(it, n))
    if len(result) == n:
        yield result
    for elem in it:
        result = result[1:] + (elem,)
        yield result


for t in window(clean_corpus.split(), n = 3):
    print(t)
```

```python
def get_prob_distribution(stem_counts, stem):
    """
    Calculates the probability distribution of the next word given a stem.

    Args:
        stem_counts (dict): A dictionary where keys are N-grams (tuples of words)
                            and values are their frequencies.
                            For example, if the stem is ('it', 'was', 'the'),
                            stem_counts should contain entries like
                            (('it', 'was', 'the'), count) and
                            (('it', 'was', 'the', 'age'), count).
        stem (tuple): The preceding sequence of words (N-gram) for which
                      to calculate the probability of the next word.

    Returns:
        dict: A dictionary where keys are the possible next words and values
              are their probabilities.
    """
    # The denominator is the count of the stem itself.
    # This represents the total occurrences of the given stem.
    denominator = stem_counts[stem]

    # Initialize an empty dictionary to store probabilities of next words
    probs = {}

    # Iterate through all N-grams and their counts in stem_counts
    for k, v in stem_counts.items():
        # Check if the current N-gram 'k' starts with the given 'stem'
        # k[:-1] means all elements of tuple k except the last one.
        # This effectively checks if 'k' is a (stem + next_word) sequence.
        if k[:-1] == stem:
            # If it matches, the last word of 'k' is the next word.
            # Calculate its probability by dividing its count (stem_counts[k])
            # by the total count of the stem (denominator).
            probs[k[-1]] = stem_counts[k] / denominator

    return probs
```

```python
stem_to_check = ('it', 'was', 'the')
probs = get_prob_distribution(stem_counts, stem_to_check)

print(f"Probability distribution for stem {stem_to_check}:")
pprint.pprint(probs)
```
